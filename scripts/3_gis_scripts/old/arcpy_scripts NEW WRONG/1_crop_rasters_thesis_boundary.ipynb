```python
import arcpy
from arcpy.sa import ExtractByMask

# check out Spatial Analyst license
arcpy.CheckOutExtension("Spatial")
```

```python
# ── PATHS ────────────────────────────────────────────────────────────────────

# Input * TEMPLATE UNEDITED * bathymetry rasters
GBC_GDB  = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\GEBCO_Gridded2025_15arcsec.gdb"
USC_GDB  = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\USCaribbean_Jeremiah_10meter.gdb"

# Input rasters (* read-only source GDBs – never written to)
GBC_bathy = GBC_GDB + r"\GBC15a_bathy"
GBC_slope = GBC_GDB + r"\GBC15a_slope"
USC_bathy = USC_GDB + r"\USC10m_bathy"
USC_slope = USC_GDB + r"\USC10m_slope"

# Input mask polygon (* must already exist in your project or a GDB you own)
boundary = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\thesis_custom_rasters.gdb\boundary_thesis"

# Output GDB (write all results here, NOT to source GDBs)
out_gdb  = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\thesis_custom_rasters.gdb"
```

```python
# ── SETTINGS ─────────────────────────────────────────────────────────────────
arcpy.env.overwriteOutput = True    # overwrite existing outputs on rerun
arcpy.env.compression = "NONE"
arcpy.env.workspace = out_gdb
```

```python
# ── HELPER FUNCTION ──────────────────────────────────────────────────────────
def extract(in_raster, mask, out_name, snap_raster, label=""):
    
    #༝ snap output raster to original input raster defined in function call chunks below
    arcpy.env.snapRaster = snap_raster
    print(f"  Extracting: {label} → {out_name}")
    
    #❥ run 'Extract by Mask' tool
    result = ExtractByMask(in_raster, mask, "INSIDE")
    
    # save result to ~\Bathymetry_geodatabases\thesis_custom_rasters.gdb
    result.save(out_gdb + "\\" + out_name)
    print(f"  Saved: {out_gdb}\\{out_name}")
```

```python
# ── 1. CROP: GEBCO BATHYMETRY RASTER ─────────────────────────────────────────
print("=== GEBCO: Bathymetry ===")
extract(
    in_raster   = GBC_bathy,
    mask        = boundary,
    out_name    = "GBC15a_bathy_thesis",
    #༝ snap to original input raster before first cut
    snap_raster = GBC_bathy,
    label       = "GBC15a_bathy → thesis"
)
```

```python
# ── 2. CROP: GEBCO SLOPE RASTER ──────────────────────────────────────────────
print("=== GEBCO: Slope ===")
extract(
    in_raster   = GBC_slope,
    mask        = boundary,
    out_name    = "GBC15a_slope_thesis",
    #༝ snap to original input raster before first cut
    snap_raster = GBC_slope,
    label       = "GBC15a_slope → thesis"
)
```

```python
# ── 3. CROP: USC BATHYMETRY RASTER ───────────────────────────────────────────
print("=== USC: Bathymetry ===")
extract(
    in_raster   = USC_bathy,
    mask        = boundary,
    out_name    = "USC10m_bathy_thesis",
    #༝ snap to original input raster before first cut
    snap_raster = USC_bathy,
    label       = "USC10m_bathy → thesis"
)
```

```python
# ── 4. CROP: USC SLOPE RASTER ────────────────────────────────────────────────
print("=== USC: Slope ===")
extract(
    in_raster   = USC_slope,
    mask        = boundary,
    out_name    = "USC10m_slope_thesis",
    #༝ snap to original input raster before first cut
    snap_raster = USC_slope,
    label       = "USC10m_slope → thesis"
)
```

```python
# ── ADD ALL CROPPED RASTERS TO CURRENT MAP ───────────────────────────────────
print("=== Adding layers to map ===")

#⊹ should be working in current map (map_setup_thesis)
aprx = arcpy.mp.ArcGISProject("CURRENT")
m    = aprx.activeMap

#༝ extracted output rasters
for out_name in [
    "GBC15a_bathy_thesis",
    "GBC15a_slope_thesis",
    "USC10m_bathy_thesis",
    "USC10m_slope_thesis"
]:
    m.addDataFromPath(out_gdb + "\\" + out_name)
    print(f"  Added output: {out_name}")

# release Spatial Analyst license
arcpy.CheckInExtension("Spatial")

print("=== Done ===")
```
