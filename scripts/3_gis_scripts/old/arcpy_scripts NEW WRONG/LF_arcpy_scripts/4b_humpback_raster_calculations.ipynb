```python
# ── LOAD DATA FOR MISSION ────────────────────────────────────────────────────
points = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Missions\arcgis_UVI2_Around_STX_20250306\arcgis_UVI2_Around_STX_20250306.gdb\calc_humpback_points_NAD83"

bathy_raster = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\custom_rasters.gdb\MERGED_BATHY_THESIS"
slope_raster = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\custom_rasters.gdb\MERGED_SLOPE_THESIS"

# final field names
depth_final = "RASTERDEPTH_m"
slope_final = "RASTERSLOPE_deg"
```

```python
# ── SETTINGS ─────────────────────────────────────────────────────────────────
arcpy.env.overwriteOutput = True
```

```python
# ── HELPER ────────────────────────────────────────────────────────────────────
def add_field_if_missing(table, field_name, field_type, field_length=None):
    existing = [f.name for f in arcpy.ListFields(table)]
    if field_name not in existing:
        if field_type.upper() == "TEXT" and field_length is not None:
            arcpy.management.AddField(table, field_name, field_type, field_length=field_length)
        else:
            arcpy.management.AddField(table, field_name, field_type)
```

```python
# ── EXTRACT MULTI VALUES TO POINTS (bilinear interpolation) ───────────────────
ExtractMultiValuesToPoints(
    in_point_features=points,
    in_rasters=[
        [bathy_raster, depth_final],
        [slope_raster, slope_final],
    ],
    bilinear_interpolate_values="BILINEAR"
)

print("Done.")
```
