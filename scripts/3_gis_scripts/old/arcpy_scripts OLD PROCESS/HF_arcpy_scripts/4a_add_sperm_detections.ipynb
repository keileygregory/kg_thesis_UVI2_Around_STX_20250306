```python
import arcpy
```

```python
# ── PATHS ──────────────────────────────────────────────────────────────────────
csv_in     = r"C:\Users\kkeii\OneDrive\Documents\THESIS_REPOSITORIES\MISSION_REPOSITORIES\UVI2_Around_STX_20250306\data\4_data_interpolated\sperm_INTERPOLATED_UVI2_Around_STX_20250306.csv"

gdb        = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Projects\THESIS_PROJECTS\map_sperm_UVI2_Around_STX_20250306\map_sperm_UVI2_Around_STX_20250306.gdb"
aprx_path  = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Projects\THESIS_PROJECTS\map_sperm_UVI2_Around_STX_20250306\map_sperm_UVI2_Around_STX_20250306.aprx"

heatmap_wgs84 = gdb + r"\sperm_heatmap_WGS84"
heatmap_nad83 = gdb + r"\sperm_heatmap_NAD83"
pts_wgs84 = gdb + r"\sperm_points_WGS84"
pts_nad83 = gdb + r"\calc_sperm_points_NAD83"

geog_transform = "WGS_1984_(ITRF00)_To_NAD_1983"

aprx       = arcpy.mp.ArcGISProject("CURRENT")
active_map = aprx.activeMap
```

```python
# ── 1. ADD CSV TO MAP ──────────────────────────────────────────────────────────
csv_layer = active_map.addDataFromPath(csv_in)
print(f"Added to map: {csv_layer.name}")
```

```python
# ── 2. XY TABLE TO POINT (heatmap layer) ───────────────────────────────
arcpy.management.XYTableToPoint(
    in_table          = csv_layer,
    out_feature_class = heatmap_wgs84,
    x_field           = "av_longitude_DD",
    y_field           = "av_latitude_DD",
    z_field           = None,
    coordinate_system = arcpy.SpatialReference(4326)
)
print(f"Created: {heatmap_wgs84}")
```

```python
# ── 3. PROJECT HEATMAP ───────────────────────────────────────────────────
arcpy.management.Project(
    in_dataset              = heatmap_wgs84,
    out_dataset             = heatmap_nad83,
    out_coor_system         = arcpy.SpatialReference("NAD 1983 UTM Zone 20N"),
    transform_method        = geog_transform,
    in_coor_system          = arcpy.SpatialReference(4326),
    preserve_shape          = "NO_PRESERVE_SHAPE"
)
print(f"Projected: {heatmap_nad83}")
```

```python
# ── 4. XY TABLE TO POINT (invisible calc layer) ───────────────────────────────
arcpy.management.XYTableToPoint(
    in_table          = csv_layer,
    out_feature_class = pts_wgs84,
    x_field           = "av_longitude_DD",
    y_field           = "av_latitude_DD",
    z_field           = None,
    coordinate_system = arcpy.SpatialReference(4326)
)
print(f"Created: {pts_wgs84}")
```

```python
# ── 5. PROJECT POINTS ──────────────────────────────────────────────────────────
arcpy.management.Project(
    in_dataset              = pts_wgs84,
    out_dataset             = pts_nad83,
    out_coor_system         = arcpy.SpatialReference("NAD 1983 UTM Zone 20N"),
    transform_method        = geog_transform,
    in_coor_system          = arcpy.SpatialReference(4326),
    preserve_shape          = "NO_PRESERVE_SHAPE"
)
print(f"Projected: {pts_nad83}")
```
