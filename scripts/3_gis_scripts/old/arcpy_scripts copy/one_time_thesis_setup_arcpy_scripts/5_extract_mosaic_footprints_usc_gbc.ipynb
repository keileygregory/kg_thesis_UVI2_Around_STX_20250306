```python
# ── PATHS ────────────────────────────────────────────────────────────────────────────────────────────

# Input mosaic datasets
mosaic_bathy = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\MASTER_RASTERS_USC_GBC_MERGED.gdb\mosaic_bathy_USC10mGBC15a_p26920_v6642"
mosaic_slope = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\MASTER_RASTERS_USC_GBC_MERGED.gdb\mosaic_slope_USC10mGBC15a_p26920_v6642"

# Ouput footprint polygons
footprint_bathy = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\MASTER_RASTERS_USC_GBC_MERGED.gdb\footprint_mosaic_bathy_p26920_v6642"
footprint_slope = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\MASTER_RASTERS_USC_GBC_MERGED.gdb\footprint_mosaic_slope_p26920_v6642"

usc_footprint_bathy = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\MASTER_RASTERS_USC_GBC_MERGED.gdb\USC_footprint_mosaic_bathy_p26920_v6642"
gbc_footprint_bathy = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\MASTER_RASTERS_USC_GBC_MERGED.gdb\GBC_footprint_mosaic_bathy_p26920_v6642"

usc_footprint_slope = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\MASTER_RASTERS_USC_GBC_MERGED.gdb\USC_footprint_mosaic_slope_p26920_v6642"
gbc_footprint_slope = r"C:\Users\kkeii\OneDrive\Documents\ArcGIS\Files\THESIS_FILES\Geodatabases\Bathymetry_geodatabases\MASTER_RASTERS_USC_GBC_MERGED.gdb\GBC_footprint_mosaic_slope_p26920_v6642"
```

```python
# ── SETTINGS ─────────────────────────────────────────────────────────────────────────────────────────
arcpy.env.overwriteOutput = True    # overwrite existing outputs on rerun
```

```python
# ── 1. EXTRACT FOOTPRINT POLYGON FEATURE CLASS FROM MOSAICS ──────────────────────────────────────────

# Export Mosaic Dataset Geometry (Data Management Tools): Bathymetry
arcpy.management.ExportMosaicDatasetGeometry(
    in_mosaic_dataset=mosaic_bathy, 
    out_feature_class=footprint_bathy, 
    geometry_type="FOOTPRINT"
)

# Export Mosaic Dataset Geometry (Data Management Tools): Slope
arcpy.management.ExportMosaicDatasetGeometry(
    in_mosaic_dataset=mosaic_slope, 
    out_feature_class=footprint_slope, 
    geometry_type="FOOTPRINT"
)

# Print values in 'Name' column of each output footprint Attribute Table to confirm "where_clause" strings below
for fc in (footprint_bathy, footprint_slope):
    print(fc.split("\\")[-1])
    with arcpy.da.SearchCursor(fc, ["Name"]) as cursor:
        for row in cursor:
            print(f"    {row[0]}")
```

```python
# ── 2. EXTRACT INDIVIDUAL INPUT RASTER (USC/GBC) BOUNDARY POLYGONS FROM FOOTPRINTS ───────────────────

# Select (Analysis Tools): Bathymetry mosaic footprint - GBC boundary
arcpy.analysis.Select(
    in_features=footprint_bathy, 
    out_feature_class=gbc_footprint_bathy, 
    # select row where 'Name' col has string value "nooverlap_GBC15a_bathy_p26920_v6642"
    where_clause="Name = 'nooverlap_GBC15a_bathy_p26920_v6642'"
)

# Select (Analysis Tools): Bathymetry mosaic footprint - USC boundary
arcpy.analysis.Select(
    in_features=footprint_bathy, 
    out_feature_class=usc_footprint_bathy, 
    # select row where 'Name' col has string value "USC10m_bathy_p26920_v6642"
    where_clause="Name = 'USC10m_bathy_p26920_v6642'"
)

# Select (Analysis Tools): Slope mosaic footprint - GBC boundary
arcpy.analysis.Select(
    in_features=footprint_slope, 
    out_feature_class=gbc_footprint_slope, 
    # select row where 'Name' col has string value "nooverlap_GBC15a_slope_p26920_v6642"
    where_clause="Name = 'nooverlap_GBC15a_slope_p26920_v6642'"
)

# Select (Analysis Tools): Slope mosaic footprint - USC boundary
arcpy.analysis.Select(
    in_features=footprint_slope, 
    out_feature_class=usc_footprint_slope, 
    # select row where 'Name' col has string value "USC10m_slope_p26920_v6642"
    where_clause="Name = 'USC10m_slope_p26920_v6642'"
)
```
