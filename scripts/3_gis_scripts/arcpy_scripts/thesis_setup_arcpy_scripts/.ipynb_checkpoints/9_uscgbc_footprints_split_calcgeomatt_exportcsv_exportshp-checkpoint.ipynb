```python
# ── SETTINGS ─────────────────────────────────────────────────────────────────────────────────────────
arcpy.env.overwriteOutput = True    # overwrite existing outputs on rerun
```

```python
# ── 1. EXTRACT FOOTPRINT POLYGON FEATURE CLASSES FROM MOSAICS ────────────────────────────────────────

# Export Mosaic Dataset Geometry (Data Management Tools): Function
def export_mosaic_footprint(md, md_footprint):

    # Export Mosaic Dataset Geometry (Data Management Tools):
    try:
        arcpy.management.ExportMosaicDatasetGeometry(
            in_mosaic_dataset=md, 
            out_feature_class=md_footprint, 
            geometry_type="FOOTPRINT"
        )
        print(f"✅ Created polygon feature class of {os.path.basename(md)} mosaic footprint: {md_footprint} \n")

    # print warning if tool failed
    except Exception as e:
        print(f"❌[FAILURE] Footprint extraction failed for {os.path.basename(md)}: {e} \n")


# Call function for bathy and slope mosaic datasets
export_mosaic_footprint(md_bathy, footprint_bathy)
export_mosaic_footprint(md_slope, footprint_slope)
```

```python
# ── GET NAMES FOR EXTRACTING INDIVIDUAL INPUT RASTER (USC/GBC) BOUNDARY POLYGONS FROM FOOTPRINTS ─────

# Print values in 'Name' column of each output mosaic footprint Attribute Table to confirm "where_clause" strings below
for fc in (footprint_bathy, footprint_slope):
    print(fc.split("\\")[-1])
    with arcpy.da.SearchCursor(fc, ["Name"]) as cursor:
        for row in cursor:
            print(f"    {row[0]}")
```

```python
# ── 2. EXTRACT INDIVIDUAL INPUT RASTER (USC/GBC) BOUNDARY POLYGONS FROM FOOTPRINTS ───────────────────

# Select (Analysis Tools): Function
def select_footprint(footprint, output_split_footprint, name_value):

    try:
        arcpy.analysis.Select(
            in_features=footprint, 
            out_feature_class=output_split_footprint, 
            # select row where 'Name' col has string value matching name_value
            where_clause=f"Name = '{name_value}'"
        )
        print(f"✅ Extracted individual USC/GBC footprint: {output_split_footprint} \n")

    # print warning if tool failed
    except Exception as e:
        print(f"❌[FAILURE] USC/GBC footprint extraction failed for {os.path.basename(output_split_footprint)}: {e} \n")


# Call function for individual GBC/USC input raster footprints in each mosaic
select_footprint(footprint_bathy, gbc_footprint_bathy, "nooverlap_GBC15a_bathy_p26920_v6642")
select_footprint(footprint_bathy, usc_footprint_bathy, "USC10m_bathy_p26920_v6642")
select_footprint(footprint_slope, gbc_footprint_slope, "nooverlap_GBC15a_slope_p26920_v6642")
select_footprint(footprint_slope, usc_footprint_slope, "USC10m_slope_p26920_v6642")
```

```python
# ── INSPECT EXTRACTED INDIVIDUAL USC/GBC FOOTPRINT FEATURE CLASSES ───────────────────────────────────

# Function to print info (geometry type, feature count, coordinate systems) for each feature class
def print_fc_info(fc):

    # get geometry type/coordinate system info
    d = arcpy.Describe(fc)
    # get feature count info
    count = arcpy.management.GetCount(fc)[0]
    # get spatial reference info
    sr = d.spatialReference
    vcs = sr.VCS.name if sr.VCS else "NONE"

    # info print block for feature class
    print(f"\n---------------- {os.path.basename(fc)} ----------------: \n")
    print(f"   Geometry type: {d.shapeType}")
    print(f"   Feature count: {count}")
    print(f"   Horizontal SR: {sr.name}")
    print(f"   Vertical SR: {vcs}")
    print(f"\n")


# Call function to print info for each extracted USC/GBC footprint feature class
print_fc_info(gbc_footprint_bathy)
print_fc_info(usc_footprint_bathy)
print_fc_info(gbc_footprint_slope)
print_fc_info(usc_footprint_slope)
```

```python
# ── 3. ADD COLUMNS TO ATTRIBUTE TABLES WITH MIN/MAX LON AND LAT ──────────────────────────────────────

# Calculate Geometry Attributes (Data Management Tools): Function
def add_extent_coords(fc):

    try:
        arcpy.management.CalculateGeometryAttributes(
            in_features=fc,
            geometry_property=[
            #   ["desired field name",  "calculated property"]
                ["x_coordinate_min_dd", "EXTENT_MIN_X"],
                ["x_coordinate_max_dd", "EXTENT_MAX_X"],
                ["y_coordinate_min_dd", "EXTENT_MIN_Y"],
                ["y_coordinate_max_dd", "EXTENT_MAX_Y"],
                ["area_geodesic_m2",     "AREA_GEODESIC"]
            ],
            # output geodesic area of polygon in square meters
            area_unit="SQUARE_METERS",
            # defaults to CS of input feature class (** projected EPSG:26920 will output min/max x and y coordinates in meters Easting/Northing which are not very easily interpretable so set CS for coordinates/area calculations to input polygon's *geographic* CS (EPSG:4269) to output coordinate values in decimal degrees without changing datums = *NO* reprojection needed)
            coordinate_system=arcpy.SpatialReference(4269)
            # coordinate_format leave as default ("SAME_AS_INPUT") since set coordinate_system to geographic CS output values will default to DD
        )
        print(f"✅ Calculate Geometry Attributes tool ran successfully for {os.path.basename(fc)} \n")

    # print warning if tool failed
    except Exception as e:
        print(f"❌[FAILURE] Calculate Geometry Attributes tool failed for {os.path.basename(fc)}: {e} \n")


# Call function for each extracted USC/GBC footprint feature class
add_extent_coords(gbc_footprint_bathy)
add_extent_coords(usc_footprint_bathy)
add_extent_coords(gbc_footprint_slope)
add_extent_coords(usc_footprint_slope)
```

```python
# ── INSPECT NEW FIELDS IN ATTRIBUTE TABLES ───────────────────────────────────────────────────────────

def confirm_fields_populated(fc):

    # read data from Attribute Table for fields created by Calculate Geometry Attributes tool to confirm all values populated successfully
    fields = ["x_coordinate_min_dd", "x_coordinate_max_dd", "y_coordinate_min_dd", "y_coordinate_max_dd", "area_geodesic_m2"]

    # grab values in the single row of feature class Attribute Table
    with arcpy.da.SearchCursor(fc, fields) as cursor:
        row = next(cursor)

    # print message with field names and values, or warning message if any field returned null
    if all(v is not None for v in row):
        print(f"\n✅ All fields successfully populated for {os.path.basename(fc)}: \n")
        for name, val in zip(fields, row):
            print(f"  '{name}': {val}")
    else:
        print(f"\n⚠️[WARNING] One or more fields are null for {os.path.basename(fc)}: \n")
        for name, val in zip(fields, row):
            print(f"  '{name}': {val}")


# Call function for Attribute Table of each extracted USC/GBC footprint feature class
confirm_fields_populated(gbc_footprint_bathy)
confirm_fields_populated(usc_footprint_bathy)
confirm_fields_populated(gbc_footprint_slope)
confirm_fields_populated(usc_footprint_slope)
```

```python
# ── 4. EXPORT ATTRIBUTE TABLES TO CSVS ───────────────────────────────────────────────────────────────

# Export Table (Conversion Tools): Function
def export_table_to_csv(fc, out_csv):

    try:
        arcpy.conversion.ExportTable(
            in_table=fc,
            out_table=out_csv
        )
        print(f"✅ Attribute Table data exported to CSV: {out_csv} \n")

    # print warning if tool failed
    except Exception as e:
        print(f"❌[FAILURE] CSV export failed for {os.path.basename(out_csv)}: {e} \n")


# Call function for Attribute Table of each extracted USC/GBC footprint feature class
export_table_to_csv(gbc_footprint_bathy, csv_gbc_footprint_bathy)
export_table_to_csv(usc_footprint_bathy, csv_usc_footprint_bathy)
export_table_to_csv(gbc_footprint_slope, csv_gbc_footprint_slope)
export_table_to_csv(usc_footprint_slope, csv_usc_footprint_slope)
```

```python
# ── 5. EXPORT INDIVIDUAL USC/GBC FOOTPRINT FEATURE CLASSES AS SHAPEFILES ─────────────────────────────

# Export Features (Conversion Tools): Function
def export_fc_to_shp(in_fc, out_shp):

    try:
        arcpy.conversion.ExportFeatures(
            in_features=in_fc,
            out_features=out_shp
        )
        print(f"✅ Exported shapefile: {out_shp} \n")

    # print warning if tool failed
    except Exception as e:
        print(f"❌[FAILURE] Shapefile failed to export for {os.path.basename(out_shp)}: {e} \n")


# Call function for each individual GBC/USC footprint polygon feature classes
export_fc_to_shp(gbc_footprint_bathy, shp_gbc_footprint_bathy)
export_fc_to_shp(usc_footprint_bathy, shp_usc_footprint_bathy)
export_fc_to_shp(gbc_footprint_slope, shp_gbc_footprint_slope)
export_fc_to_shp(usc_footprint_slope, shp_usc_footprint_slope)
```
